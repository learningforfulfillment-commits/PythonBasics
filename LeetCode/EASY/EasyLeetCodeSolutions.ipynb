```python
# 242 Valid Anagram
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s)!= len(t):
            return False
        countS, countT = {},{}
        # create hashmaps
        for i in range(len(s)):
            countS[s[i]] = 1 + countS.get(s[i],0)
            countT[t[i]] = 1+ countT.get(t[i],0)
        # compare hashmaps
        for c in countS: # iterate through keys. c is the key
            if countS[c] != countT.get(c,0): # if count at character c is same in s & t
                return False
        return True
        # return Counter(s) == Counter(t)
        # Time : O(s+t)
        # Space: O(n)

        # return sorted(s) == sorted(t)
        # sorting Time : O(nlogn), O(n**2)
        # sorting space : O(1),O(n)
        
```

```python
# 1. Two Sum
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i, num in enumerate(nums):  # i = index, num = value
            diff = target - num
            if diff in seen:
                return [seen[diff], i]  # return the index of the pair
            # saving the number as a key and its position (index) as the value inside the dictionary seen.
            seen[num] = i # Remember that I saw the number num at index i

"""
Approach    Code	        Time	Space	Notes
BruteForce	Double loop	    O(n²)	O(1)	Simple but slow
Hash Map	seen[num] = i	O(n)	O(n)	Fastest and most common
"""

```

```python
# 1260 - Shift 2D Grid
class Solution:
    def shiftGrid(self, grid: List[List[int]], k: int) -> List[List[int]]:# grid is a list of lists.Grid is box of rows
#     grid = [
#     [1, 2, 3],
#     [4, 5, 6]
# ]
# row= 2
# column = 3
# METHOD: Linearization of a 2D array  --> Flatten → Rotate → Rebuild
# Array Rotation , Simulation, Flattening all follow same method  --> LC 48,54,73
# 1️. Take all numbers out → one line
# 2️. Move the line → shift
# 3️. Put numbers back → grid

        rows = len(grid)
        cols = len(grid[0])

        # Step 1: Flatten the grid -->Visits every number once. Time = m × n
        #flat = [num for row in grid for num in row] # List Comprehension for 1st step
        flat = [] # flat list → stores all m × n elements
        for row in grid:
            for num in row:
                flat.append(num)

        # Step 2: Rotate the list
        k = k % len(flat)
        flat = flat[-k:] + flat[:-k] # Give me the last k toys + Give me everything EXCEPT the last k toys. it is just cut and paste  --> Copies all numbers once. Time = m × n
        # example if k= 2 ,flat[-2:]  → [3, 4] and flat[:-2]  → [1, 2] and flate = [3, 4, 1, 2]

        # Step 3: Build the new grid
        result = [] # new grid --> result grid → stores all m × n elements
        index = 0 # points at flat list
        for i in range(rows): # For each row…
            row = [] # Create one row
            for j in range(cols): # For each column ...
                row.append(flat[index])#Take number from flat list   --> Again visits every number once. Time = m × n
                index += 1# Finger moves right
            result.append(row)

        return result
        #  return [flat[i * column:(i + 1) * column] for i in range(rows)] # List Comprehension for 3rd step

# | index | take | row     |
# | ----- | ---- | ------- |
# | 0     | 6    | [6]     |
# | 1     | 1    | [6,1]   |
# | 2     | 2    | [6,1,2] |

# T : O(m*n)
# S : O(m*n)
```
