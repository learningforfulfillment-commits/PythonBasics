```python
# Jan 19, 2026 - LC 367. Valid Perfect Square
# --------------------------------------------------------
# https://leetcode.com/problems/valid-perfect-square/description/
# Math ,Binary Search
# A number is a perfect square if: You can multiply a whole number by itself and get that number.
# 1 + 3 + 5 + 7 + ... = perfect square
class Solution:
    def isPerfectSquare(self, num: int) -> bool:
        if num < 2:  # 0 and 1 are perfect squares
            return True

        left, right = 1, num  # always search from 1 to num

        while left <= right:
            mid = left + (right - left) // 2
            square = mid * mid

            if square == num:
                return True
            elif square < num:
                left = mid + 1
            else:
                right = mid - 1

        return False


# T : O(log n)
# S : O(1)

```
