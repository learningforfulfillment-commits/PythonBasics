```python
# Jan 18, 2026 -  Binary Search (LeetCode 704)
# Array, Binary Search
# Binary Search is an efficient searching algorithm used to find an element in a sorted array.
# Repeatedly divides the search space in half
# Decides which half to keep searching based on comparisons
# This makes it very fast.

# Binary Search works ONLY if:
#  The array is sorted
#  You can access elements by index (arrays / lists)

#  O(log n) Time Complexity?
# Each step cuts the array into half.
# Linear Search → O(n)
# Binary Search → O(log n)
class Solution:
    def search(self, nums: List[int], target: int) -> int:

        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return mid
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        return -1

# Time Complexity	O(log n)
# Space	O(1) iterative

    # def binary_search(nums, left, right, target):
    # METHOD 2 - RECURSIVE
    #     if left > right:
    #         return -1

    #     mid = (left + right) // 2

    #     if nums[mid] == target:
    #         return mid
    #     elif nums[mid] < target:
    #         return binary_search(nums, mid + 1, right, target)
    #     else:
    #         return binary_search(nums, left, mid - 1, target)


# def first_occurrence(nums, target):
# Array is sorted but duplicates exist.Even after finding target, don’t stop.Move:Left to find first and Right to find last
#     left, right = 0, len(nums) - 1
#     ans = -1

#     while left <= right:
#         mid = (left + right) // 2

#         if nums[mid] == target:
#             ans = mid
#             right = mid - 1   # move left
#         elif nums[mid] < target:
#             left = mid + 1
#         else:
#             right = mid - 1

#     return ans

# def last_occurrence(nums, target):
#     left, right = 0, len(nums) - 1
#     ans = -1

#     while left <= right:
#         mid = (left + right) // 2

#         if nums[mid] == target:
#             ans = mid
#             left = mid + 1   # move right
#         elif nums[mid] < target:
#             left = mid + 1
#         else:
#             right = mid - 1

#     return ans

# def lower_bound(nums, target):
#     left, right = 0, len(nums)

#     while left < right:
#         mid = (left + right) // 2

#         if nums[mid] < target:
#             left = mid + 1
#         else:
#             right = mid

#     return left

# def upper_bound(nums, target):
#     left, right = 0, len(nums)

#     while left < right:
#         mid = (left + right) // 2

#         if nums[mid] <= target:
#             left = mid + 1
#         else:
#             right = mid

#     return left


# def search_rotated(nums, target):
#     left, right = 0, len(nums) - 1

#     while left <= right:
#         mid = (left + right) // 2

#         if nums[mid] == target:
#             return mid

#         # left half sorted
#         if nums[left] <= nums[mid]:
#             if nums[left] <= target < nums[mid]:
#                 right = mid - 1
#             else:
#                 left = mid + 1
#         else:
#             # right half sorted
#             if nums[mid] < target <= nums[right]:
#                 left = mid + 1
#             else:
#                 right = mid - 1

#     return -1


# Real-World Applications 
# Binary Search is everywhere:
    #  Software:
        # Database indexing
        # Search engines
        # Auto-complete
        # Version control (first bad commit)
    # DSA Problems:
        # Scheduling
        # Allocation
        # Optimization
        # Capacity planning

# | Type          | Used For        |
# | ------------- | --------------- |
# | Basic BS      | Find element    |
# | First/Last    | Duplicates      |
# | Lower/Upper   | Ranges          |
# | Answer Space  | Optimization    |
# | Rotated Array | Modified sorted |
```

```python
# Jan 18, 2026 - LC 496. Next Greater Element I
# Array, Hash Table, Stack, Monotonic Stack

# A monotonic stack is a stack that maintains elements in: Increasing order, or Decreasing order
# For Next Greater Element: We use a monotonic decreasing stack
class Solution:
    def nextGreaterElement(self, nums1: List[int], nums2: List[int]) -> List[int]:
        stack = []
        next_greater = {}

        # Step 1: Process nums2
        for num in nums2:
            while stack and num > stack[-1]:
                prev = stack.pop()
                next_greater[prev] = num
            stack.append(num)

        # Step 2: Remaining elements have no next greater
        while stack:
            next_greater[stack.pop()] = -1

        # Step 3: Build result for nums1
        return [next_greater[num] for num in nums1]

# Time	O(n + m)
# Space	O(n)
# Each element:Pushed once, Popped once
# Why Stack Is Perfect Here  : Nearest greater, To the right, Efficient
# Stack gives: Order, Memory, One-pass solution
# Next Greater, Next Smaller, Nearest Greater, Nearest Smaller -→ Think Monotonic Stack immediately.
# | Problem            | Stack Type |
# | ------------------ | ---------- |
# | Next Greater Right | Decreasing |
# | Next Greater Left  | Decreasing |
# | Next Smaller Right | Increasing |
# | Stock Span         | Decreasing |
# | Daily Temperatures | Decreasing |

# Real-World Applications 
    # Stock price analysis
    # Temperature forecasting
    # Histogram problems
    # Scheduling systems
    # CPU task optimization

```

```python
# Jan 18, 2026 - LC 94. Binary Tree Inorder Traversal
# Stack ,Tree, Depth-First Search, Binary Tree

# | Traversal | Order               |
# | --------- | ------------------- |
# | Inorder   | Left → Root → Right |
# | Preorder  | Root → Left → Right |
# | Postorder | Left → Right → Root |
# In a Binary Search Tree (BST),Inorder traversal gives: SORTED ORDER

# inorder traversal using:  Recursion (easy) or  Stack (no recursion)
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        result = []

        def dfs(node):
            if not node:
                return

            dfs(node.left)      # LEFT
            result.append(node.val)  # ROOT
            dfs(node.right)     # RIGHT

        dfs(root)
        return result

        
```
