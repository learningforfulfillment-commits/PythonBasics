```python
# Day 19 - Pattern Printer - Hollow Square Star Pattern
n = 4

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# Day 20 - Pattern Printer - Hollow Pyramid Star Pattern
n = 5

for i in range(1, n + 1):
    for j in range(1, n - i + 1):
        print(" ", end="")
    for j in range(1, 2 * i):
        if j == 1 or j == 2*i - 1 or i == n:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# Day 21 - Pattern Printer - Inverted Pyramid Star Pattern
n = 5

for i in range(n, 0, -1):
    print(" " * (n - i) + "* " * i)
```

```python
# Day 22 - Pattern Printer - Floyd's Triangle
num = 1
n = 4

for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
# Day 23 - Pattern Printer - Pascal's Triangle
n = 4

for i in range(n):
    num = 1
    print(" " * (n - i), end="")
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print()
```

```python
# Day 24 - Pattern Printer - Butterfly Star Pattern
n = 3

for i in range(1, n + 1):
    print("* " * i + "  " * (n - i) * 2 + "* " * i)

for i in range(n, 0, -1):
    print("* " * i + "  " * (n - i) * 2 + "* " * i)
```

```python
# Day 26 - Pattern Printer - X Star Pattern
n = 5

for i in range(n):
    for j in range(n):
        if i == j or i + j == n - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# Day 27 - Pattern Printer - Hollow Diamond Star Pattern
n = 4

for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(1, 2*i):
        if j == 1 or j == 2*i - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()

for i in range(n - 1, 0, -1):
    print(" " * (n - i), end="")
    for j in range(1, 2*i):
        if j == 1 or j == 2*i - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```
