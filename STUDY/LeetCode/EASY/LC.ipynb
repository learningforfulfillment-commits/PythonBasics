```python
# Jan 15, 2026 - 606. Construct String from Binary Tree
# String , Tree, Depth-First Search, Binary Tree

# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def tree2str(self, root: Optional[TreeNode]) -> str:
        # METHOD 1 - Use preorder traversal (root → left → right).
        # Recursion (depth-first) because the problem naturally fits preorder traversal.
        # Step 1: Handle the base case
        # If the tree node is None, return an empty string.
        if not root:
            return ""
        # Step 2: Start with the current node’s value
        # Every node contributes its value to the string
        # Convert integer to string for concatenation
        result = str(root.val)
        # Step 3: Process the left child
        if root.left:
            result += "(" + self.tree2str(root.left) + ")"

        # Step 4: Process the right child
        # if there’s a right child but no left child, include () to indicate missing left
        if root.right:
            if not root.left:
                result += "()"
            result += "(" + self.tree2str(root.right) + ")"

        # Step 5: Return the result
        return result
 
        # Time & Space Complexity
        # Complexity	Explanation
        # Time O(N)	Each node is visited exactly once. N = number of nodes.
        # Space O(H)	Recursion stack grows to max tree height. 
        # H = height of tree. 
        # In worst-case skewed tree: H = N. 
        # Best-case balanced tree: H = log N. 
        # Output string also uses O(N) space, so total space = O(N).

        # Recursion uses depth-first traversal, so memory depends on tree height
        # BFS would use O(width) space (can be worse for very wide trees O (N/2))
        # Skewed trees can cause deep recursion → risk of stack overflow in Python
        # Python default recursion limit ≈ 1000 → stack overflow if N > 1000
        
        # DFS Traversals
        # Preorder, inorder, postorder
        # Preorder (Root → Left → Right) - Preorder: 1 → 2 → 3 - tree-to-string, tree cloning, serialization
        # Inorder (Left → Root → Right) - Inorder: 2 → 1 → 3 - Produces sorted order for binary search trees (BST)
        # Postorder (Left → Right → Root) -- Postorder: 2 → 3 → 1 - Deleting a tree , calculating height or evaluating expressions

        # Push = put plate on top
        # Pop = take the top plate off
        # Stack is LIFO (Last In, First Out) → the last thing we push is processed first
        # We push right first → left child is on top → left is processed first (preorder: left before right)

        # Stack memory = max depth
        # Avoid stack overflow for very deep trees

        # Iterative DFS
        # Using explicit stack instead of recursion
        # Can avoid recursion depth limits

        # BFS / Level-order Traversal
        # Uses a queue
        # Memory depends on width of tree
        # Useful for printing trees level-by-level

        # DFS / Preorder recursion → memory = height of tree
        # BFS / Queue → memory = max width of tree
        # Stack overflow risk → skewed tree + deep recursion
```

```python
# Jan 15, 2026 - 682. Baseball Game
# Array, Stack, Simulation
class Solution:
    def calPoints(self, operations: List[str]) -> int:
        stack = []
        for op in operations:
            if op == 'C':
                stack.pop()
            elif op == 'D':
                stack.append(stack[-1]*2)
            elif op == '+':
                stack.append(stack[-1]+ stack[-2])
            else:
                stack.append(int(op))
            
        return sum(stack)
    # Stack Problem
    # T : O(n)
    # S : O(n)

        
```

```python
# Jan 15, 2026 - 1260. Shift 2D Grid
# Array
# Matrix
# Simulation
class Solution:
    def shiftGrid(self, grid: List[List[int]], k: int) -> List[List[int]]:# grid is a list of lists.Grid is box of rows
#     grid = [
#     [1, 2, 3],
#     [4, 5, 6]
# ]
# row= 2
# column = 3

        rows = len(grid)
        cols = len(grid[0])

        # Step 1: Flatten the grid
        flat = []
        for row in grid:
            for num in row:
                flat.append(num)

        # Step 2: Rotate the list
        k = k % len(flat)
        flat = flat[-k:] + flat[:-k] # Give me the last k toys + Give me everything EXCEPT the last k toys. it is just cut and paste
        # example if k= 2 ,flat[-2:]  → [3, 4] and flat[:-2]  → [1, 2] and flate = [3, 4, 1, 2]

        # Step 3: Build the new grid
        result = []
        index = 0
        for i in range(rows): # For each row…
            row = [] # Create one row
            for j in range(cols): # For each column ...
                row.append(flat[index])#Take number from flat list
                index += 1# Finger moves right
            result.append(row)

        return result

# | index | take | row     |
# | ----- | ---- | ------- |
# | 0     | 6    | [6]     |
# | 1     | 1    | [6,1]   |
# | 2     | 2    | [6,1,2] |

```

```python
# Jan 15, 2026 - 70. Climbing Stairs
# Math
# Dynamic Programming
# Memoization
class Solution:
    def climbStairs(self, n: int) -> int:
        memory_1, memory_2 = 1,1
        for i in range(n-1):
            temp = memory_1
            memory_1 = memory_1 + memory_2
            memory_2 = temp
        return memory_1
# T : O(n)
# S : O(1)
```

```python
# Jan 15, 2026 - 198. House Robber
# Array ,Dynamic Programming
class Solution:
    def rob(self, nums: List[int]) -> int:
        moshanam2, moshanam1 = 0,0
        # [...,moshanam2,moshanam1,n,n+1,n+2,...]
        for n in nums:
            temp = max(moshanam2 + n, moshanam1)
            moshanam2 = moshanam1
            moshanam1 = temp
        return moshanam1

    # T: O(n)
    # S: O(n)
        
```

```python
# Jan 15, 2026 - 242. Valid Anagram
# Hash Table
# String
# Sorting
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s)!= len(t):
            return False
        countS, countT = {},{}
        # create hashmaps
        for i in range(len(s)):
            countS[s[i]] = 1 + countS.get(s[i],0)
            countT[t[i]] = 1+ countT.get(t[i],0)
        # compare hashmaps
        for c in countS: # iterate through keys. c is the key
            if countS[c] != countT.get(c,0): # if count at character c is same in s & t
                return False
        return True
        # return Counter(s) == Counter(t)
        # Time : O(s+t)
        # Space: O(n)

        # return sorted(s) == sorted(t)
        # sorting Time : O(nlogn), O(n**2)
        # sorting space : O(1),O(n)
        
```

```python
# Jan 15, 2026 - 20. Valid Parentheses
# String
# Stack
class Solution:
    def isValid(self, s: str) -> bool:
        stack = [] # list
        mapping = {')': '(', '}': '{', ']': '['}

        for i in s:
            # if i is closed bracket
            if i in mapping:
                top_element = stack.pop() if stack else '#'
                if mapping[i] != top_element:
                    return False
            else:
                # if i is open bracket
                stack.append(i)
        
        # Stack must be empty for valid parentheses
        return not stack
"""
Time : O(n)
Space : O(n)
"""



       
    
        
```
