```python
# Jan 20, 2026 - LC 441. Arranging Coins
# --------------------------------------------------------
# Math , Binary Search
# 1+2+3+...+k = k(k+1)​/2

class Solution:
    def arrangeCoins(self, n: int) -> int:
        left, right = 0, n

        while left <= right:
            mid = left + (right - left) // 2
            total = mid * (mid + 1) // 2

            if total == n:
                return mid
            elif total < n:
                left = mid + 1
            else:
                right = mid - 1

        return right  # largest mid with total <= n

# | Time   | O(log n) → binary search |
# | Space  | O(1) → only variables    |


        
```

```python
# Jan 20, 2026 - LC 374. Guess Number Higher or Lower
# --------------------------------------------------------
# Binary Search, Interactive

# binary Search Strategy
# Start with: left = 1 and right = n
# Guess the middle number: Use guess(mid) to decide: Too big → go LEFT, Too small → go RIGHT, Correct → return mid

class Solution:
    def guessNumber(self, n: int) -> int:
        left, right = 1, n

        while left <= right:
            mid = left + (right - left) // 2
            res = guess(mid)

            if res == 0:
                return mid
            elif res == -1:
                right = mid - 1
            else:
                left = mid + 1

# Time	O(log n)
# Space	O(1)
```

```python
# Jan 20, 2026 - LC 1189. Maximum Number of Balloons
# --------------------------------------------------------
# Hash Maps + Counting , String, Has Table
# Frequency Counting (Hash Map)
# This is a frequency-count problem where the limiting character determines the answer


class Solution:
    def maxNumberOfBalloons(self, text: str) -> int:
        count = {}

        for ch in text:
            count[ch] = count.get(ch, 0) + 1

        return min(
            count.get('b', 0),
            count.get('a', 0),
            count.get('l', 0) // 2,
            count.get('o', 0) // 2,
            count.get('n', 0)
        )


# Time	O(n)
# Space	O(1) (only 26 letters max)

# example text = "loonbbaloonbalxballpoon"

# After counting:
# b → 4
# a → 3
# l → 6
# o → 6
# n → 3

# Remember What ONE “balloon” Needs b a l l o o n
# So:
# b → 1
# a → 1
# l → 2
# o → 2
# n → 1

# Adjust Counts for l and o
# Because l and o are used twice per balloon:
# l → 6 // 2 = 3
# o → 6 // 2 = 3

# Now we have:
# b → 4
# a → 3
# l → 3
# o → 3
# n → 3

# Step 5: Take the Minimum
# min(4, 3, 3, 3, 3) = 3
# Only 3 Balloons can be formed.

"""
Pattern Recognition 
-----------------------

This problem belongs to:
    Counting problems
    Anagram-style problems
    Resource allocation
    Minimum frequency problems

Similar problems:
    Isomorphic Strings
    Anagram checks
    Word Pattern
    Ransom Note
"""
```

```python
# Jan 20, 2026 - LC 448. Find All Numbers Disappeared in an Array
# --------------------------------------------------------
# Array, Hash Tables
# We use the array itself as a hash set by marking visited indices negative.

class Solution:
    def findDisappearedNumbers(self, nums):
        # Step 1: Mark seen numbers
        for i in range(len(nums)):
            index = abs(nums[i]) - 1
            nums[index] = -abs(nums[index])

        # Step 2: Find missing numbers
        result = []
        for i in range(len(nums)):
            if nums[i] > 0:
                result.append(i + 1)

        return result


# Time	O(n)
# Space	O(1)

# Pattern Recognition
# This problem belongs to:
#     Index marking problems
#     In-place array modification
#     Cyclic sort family
#     Missing / duplicate number problems

# Related problems:
#     Find duplicate number
#     First missing positive
#     Set mismatch
```
