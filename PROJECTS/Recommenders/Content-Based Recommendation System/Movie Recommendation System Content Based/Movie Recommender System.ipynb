```python
# Types of recommender systems
#------------------------------

#  1. Content-Based Filtering
# ----------------------------------
#            - How it works: Tags - content-similarity
#                 Recommends items similar to what a user liked in the past, based on item features.
#            - Examples:
#                 Netflix – Suggests movies similar to ones you watched (same genre, actors)
#                 Spotify – “Because you listened to…” playlists
#                 Medium – Articles based on topics you read
#                 YouTube – Videos related to your watch history
#            - Pros:
#                 Personalized for each user
#                 No need for other users’ data
#            - Cons:
#                 Limited discovery (over-specialization)
#                 Requires well-defined item features

# 2. Collaborative Filtering
# ----------------------------------
#             - How it works: Interests - user-similarity
#                 Uses behavior and preferences of many users to make recommendations.
#                     a) User-Based Collaborative Filtering
#                     Finds users similar to you and recommends what they liked
#                      - Examples:
#                         Facebook – Friend suggestions (people similar users are connected to)
#                         Goodreads – Book recommendations based on users with similar tastes
#                     b) Item-Based Collaborative Filtering
#                     Recommends items similar to what you interacted with
#                      - Examples:
#                         Amazon – “Customers who bought this also bought”
#                         Netflix – “Users who watched this also watched”
#             - Pros:
#                 No need for item metadata
#                 Can discover new interests
#             - Cons:
#                 Cold start problem (new users/items)
#                 Data sparsity

# 3. Hybrid Recommender Systems
# ----------------------------------
#             - How it works:
#                 Combines content-based and collaborative filtering.
#             - Examples:
#                 Netflix – Combines viewing history + similar users
#                 Amazon – Uses product details + user behavior
#                 Spotify – Mixes listening history with global trends
#             - Pros:
#                 Reduces weaknesses of individual methods
#                 More accurate recommendations
#             - Cons:
#                 More complex to implement

# 4. Knowledge-Based Recommender Systems
# --------------------------------------
#             - How it works:
#                 Uses explicit knowledge about users and items (rules, constraints).
#             - Examples:
#                 Expedia / Booking.com – Travel packages based on budget, location, dates
#                 Car recommendation websites – Suggest cars based on price, fuel type, usage
#                 Real-estate portals – Property suggestions based on preferences
#             - Pros:
#                 No cold start problem
#                 Works without historical data
#             - Cons:
#                 Requires domain knowledge
#                 Hard to maintain

# 5. Demographic-Based Recommender Systems
# ---------------------------------------
#               -  How it works:
#                      Uses demographic information like age, gender, location.
#               -  Examples:
#                         Online ads – Targeted ads based on age group and location
#                         E-commerce websites – Regional or age-based product recommendations
#                         Streaming platforms – Kids vs adult content suggestions
#               -  Pros:
#                      Simple to implement
#               -  Cons:
#                     Stereotyping
#                     Less personalized

# 6. Context-Aware Recommender Systems
# --------------------------------------
#                 -  How it works:
#                     Incorporates context such as time, location, device, or mood.
#                 -  Examples:
#                     Google Maps – Nearby restaurants based on location and time
#                     Uber Eats / Zomato – Food suggestions based on time of day
#                     Spotify – Morning vs workout playlists
#                 - Pros:
#                     More relevant recommendations
#                 - Cons:
#                     Needs additional data collection

# 7. Deep Learning–Based Recommender Systems
#   --------------------------------------
#                 - How it works:
#                     Uses neural networks to model complex user-item interactions.
#                 - Examples:
#                     YouTube – Personalized video feed
#                     TikTok – “For You” page
#                     Instagram – Feed, Reels, and Explore suggestions
#                     LinkedIn – Job and connection recommendations
#                 - Pros:
#                     Handles large-scale and complex data
#                     High accuracy
#                 - Cons:
#                     Requires large datasets
#                     Less interpretable

# 8. Popularity-Based Recommender Systems
#   --------------------------------------
#                 - How it works:
#                     Recommends the most popular or trending items.
#                 - Examples:
#                      YouTube Trending
#                     Netflix Top 10
#                     Amazon Best Sellers
#                     App Store / Play Store Top Apps
#                 - Pros:
#                     Simple and effective for new users
#                 - Cons:
#                     Not personalized
```
